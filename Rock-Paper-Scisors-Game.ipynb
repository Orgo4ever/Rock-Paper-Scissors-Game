# Rock Paper Scissors Tournament

## Game Objective
The player competes against the computer in a rock, paper, scissors tournament. The player chooses how many rounds they want to play, and the first side to win the required number of rounds wins the tournament.

## Information Stored
- The player's name
- The number of rounds selected
- The number of wins needed
- The current round
- Player wins
- Computer wins
- Ties
- Round history

## Player Actions
The player enters their name, chooses the number of rounds, and selects rock, paper, or scissors each round.

## Win and Lose Conditions
The player wins if they reach the required number of wins before the computer. The player loses if the computer reaches the required number of wins first.

## Typical Turn
1. Display the current round.
2. Ask the player for rock, paper, or scissors.
3. Validate the player's input.
4. Generate the computer's random choice.
5. Compare both choices.
6. Update the game state.
7. Display the score.
8. Continue until the player or computer wins the tournament.

```python
import random


def get_player_name():
    """
    Ask the player for their name.
    If the player leaves the input blank, use 'Player' as the default name.
    """
    user = input("Please provide your name: ").strip()

    if user == "":
        print("No name entered. I will call you Player.")
        user = "Player"

    return user


user = get_player_name()
print(f"Welcome {user}!")
```

```python
#Establish choices
options=["rock", "paper", "scissors"]
```

```python
# Game State
game_state = {
    "current_round": 1,
    "player_wins": 0,
    "computer_wins": 0,
    "ties": 0,
    "round_history": []
}
```

```python
# Ask for number of rounds
def get_number_of_rounds():
    """This fuction determines the number of rounds in the tournament.
    It checks if the input is odd and 3 oor larger.
    """
    while True:
        try:
            rounds = int(input("How many rounds would you like to play? Please choose an odd number of at least 3: "))

            if rounds < 3:
                print("Please enter at least 3 rounds.")
            elif rounds % 2 == 0:
                print("Please select an odd number of rounds.")
            else:
                return rounds

        except ValueError:
            print("Please enter a valid whole number.")


number_of_rounds = get_number_of_rounds()
print(f"You selected {number_of_rounds} rounds.")
```

```python
# Tournament win conditions
wins_needed=number_of_rounds//2 +1
```

```python
# Computer selection
def get_computer_choice(options):
    """
    This fuction determines the computer choice
    """
    computer_choice = random.choice(options)
    return computer_choice
```

```python
winning_choices = {
    "rock": "scissors",
    "paper": "rock",
    "scissors": "paper"
}
```

```python
def get_user_choice(options):
    """This function requests an input from the user and checks if it is valid.
    """""
    while True:
        user_choice = input("Please select rock, paper, or scissors: ").lower().strip()

        if user_choice in options:
            return user_choice
        else:
            print("Invalid choice. Please choose rock, paper, or scissors.")
```

```python
def determine_winner(user_choice, computer_choice, winning_choices):
    """This fuction determines the winner of a round based on the user and computer choices.
    """
    if user_choice == computer_choice:
        return "tie"
    elif winning_choices[user_choice] == computer_choice:
        return "user"
    else:
        return "computer"
```

```python
def display_score(user, game_state):
    """This fuction determines the score after each round and displays it.
    """
    print(f"Score: {user}: {game_state['player_wins']} | Computer: {game_state['computer_wins']}")
```

```python
while game_state["player_wins"] < wins_needed and game_state["computer_wins"] < wins_needed:

    round_number = game_state["current_round"]

    print(f"\nRound {round_number}")

    user_choice = get_user_choice(options)
    computer_choice = get_computer_choice(options)

    print(f"{user} chose {user_choice}.")
    print(f"Computer chose {computer_choice}.")

    winner = determine_winner(user_choice, computer_choice, winning_choices)

    if winner == "tie":
        print("Tie match! No points awarded.")
        game_state["ties"] = game_state["ties"] + 1

    elif winner == "user":
        print(f"Congratulations {user}! You won this round!")
        game_state["player_wins"] = game_state["player_wins"] + 1
        game_state["current_round"] = game_state["current_round"] + 1

    else:
        print(f"Sorry {user}, the computer won this round.")
        game_state["computer_wins"] = game_state["computer_wins"] + 1
        game_state["current_round"] = game_state["current_round"] + 1

    game_state["round_history"].append({
        "round": round_number,
        "user_choice": user_choice,
        "computer_choice": computer_choice,
        "winner": winner
    })

    display_score(user, game_state)


if game_state["player_wins"] >= wins_needed:
    print(f"\nCongratulations {user}! You won the tournament!")
elif game_state["computer_wins"] >= wins_needed:
    print(f"\nSorry {user}, you lost the tournament. Maybe try again?")
```
