```python
# 1. Create a list of possible choices
# 2. Create a game state to track the round, player wins, and computer wins
# 3. Show welcome message
# 3.5 Ask the player how many games should be in the tournament
# 3.6 Validate that the number is odd
# 3.7 Calculate how many wins are needed to win the tournament
# 4. As a user to make a choice
# 5. Show an error message if the condition is invalid
# 6. Make computer select randomly from options
# 7. Set up conditions when the player wins/loses
# 8. Set up counters for win/losing. 
# 9.1 If player wins show win message
# 9.2 If player loses show loose message
# 9.3 If tie, show tie message
# 10.1 If win +1 to win counter
# 10.2 If loose +1 to loose counter
# 10.3 If tie prompt to try again. 
# 11
```

```python
# Establish user
user=input("please provide your name: ")
print(f"Welcome {user}! Can you beat the computer?")
```

```python
#Establish choices
options=["rock", "paper", "scissors"]
```

```python
# Game State
game_state={"current_round":1,
"player_wins":0,
"computer_wins":0}
            
```

```python
# Ask for number of rounds
number_of_rounds = int(input("How many rounds would you like to play? "))

while number_of_rounds % 2 == 0:
    print("Please select an odd number of rounds.")
    number_of_rounds = int(input("How many rounds would you like to play? "))

print("Tournament rounds selected")
```

```python
# Tournament win conditions
wins_needed=number_of_rounds//2 +1


```

```python
# Win message
win_message=(f"Congragulations {user}! You beat the computer!")
```

```python
# Loose message
lose_message=(f"sorry {user} the computer won. Try again.")
```

```python
# Tie message
tie="Tie match! No points awarded."
```

```python
# Final win message
final_win_message=(f"Congrats {user}! You won the tournament!")
```

```python
#Final Lose Message
final_lose_message=(f"Sorry {user}, you lost the tournament. Maybe try again?")
```

```python
# Computer selection
import random
computer_choice=random.choice(options)

```

```python
while game_state["player_wins"] < wins_needed and game_state["computer_wins"] < wins_needed:

    print(f"Round {game_state['current_round']}")

    user_choice = input("Please select rock, paper, or scissors: ")

    computer_choice = random.choice(options)

    if user_choice == computer_choice:
        print(tie)

    elif user_choice == "rock" and computer_choice == "scissors":
        print(win_message)
        game_state["player_wins"] = game_state["player_wins"] + 1

    elif user_choice == "scissors" and computer_choice == "paper":
        print(win_message)
        game_state["player_wins"] = game_state["player_wins"] + 1

    elif user_choice == "paper" and computer_choice == "rock":
        print(win_message)
        game_state["player_wins"] = game_state["player_wins"] + 1

    else:
        print(lose_message)
        game_state["computer_wins"] = game_state["computer_wins"] + 1

    print(f"Score: {user}: {game_state['player_wins']} | Computer: {game_state['computer_wins']}")

    game_state["current_round"] = game_state["current_round"] + 1
if game_state["player_wins"] >= wins_needed:
 print(final_win_message)
elif game_state["computer_wins"] >= wins_needed:
 print(final_lose_message)
    
    
```
