# Rock Paper Scissors Tournament
## Game Objective

The player competes against the computer in a Rock Paper Scissors tournament. By default, the game is a best-of-five tournament where the first side to win 3 rounds wins. The player can also choose a longer odd number of rounds, with 5 as the minimum.

### Information Stored
- The player's name
- The number of rounds selected
- The number of wins needed
- The current round
- Player wins
- Computer wins
- Ties
- Round history
## Player Actions

The player enters their name, chooses the number of rounds or presses Enter to use the default of 5, and selects rock, paper, or scissors each round. After the tournament ends, the player can choose whether to play again.

## Win and Lose Conditions

The player wins if they reach the required number of wins before the computer. The player loses if the computer reaches the required number of wins first.

For the default best-of-five version:

- The player wins by reaching 3 wins first.
- The player loses if the computer reaches 3 wins first.
## Typical Turn
- Display the current round.
- Ask the player for rock, paper, or scissors.
- Validate the player's input.
- Generate the computer's random choice.
- Compare both choices.
- Determine the round winner.
- Update the game state.
- Display the score.
- Continue until the player or computer wins the tournament.
- Ask the player if they want to play again.

```python
import random


def get_player_name():
    """
    Ask the player for their name.
    If the player leaves the input blank, use 'Player' as the default name.
    """
    user = input("Please provide your name: ").strip()

    if user == "":
        print("No name entered. I will call you Player.")
        user = "Player"

    return user


user = get_player_name()
print(f"Welcome {user}!")
```

```python
#Establish choices
options=["rock", "paper", "scissors"]
```

```python
# Game State
game_state = {
    "current_round": 1,
    "player_wins": 0,
    "computer_wins": 0,
    "ties": 0,
    "round_history": []
}
```

```python
# Determing the number of rounds
def get_number_of_rounds():
    """Ask the user how many rounds they want to play.
    If the user leaves the input blank, default to 5 rounds.
    The number must be an odd number and at least 5.
    """

    while True:
        rounds_input = input("How many rounds would you like to play? Press Enter for 5: ").strip()

        if rounds_input == "":
            print("No number entered. Defaulting to 5 rounds.")
            return 5

        try:
            rounds = int(rounds_input)

            if rounds < 5:
                print("Please enter a number that is at least 5.")

            elif rounds % 2 == 0:
                print("Please enter an odd number so there cannot be a tie tournament.")

            else:
                return rounds

        except ValueError:
            print("Please enter a valid whole number.")
```

```python
# Tournament win conditions
number_of_rounds = get_number_of_rounds()
wins_needed = number_of_rounds // 2 + 1
```

```python
# Computer selection
def get_computer_choice(options):
    """
    This function determines the computer choice
    """
    computer_choice = random.choice(options)
    return computer_choice
```

```python
winning_choices = {
    "rock": "scissors",
    "paper": "rock",
    "scissors": "paper"
}
```

```python
def get_user_choice(options):
    """This function requests an input from the user and checks if it is valid.
    """""
    while True:
        user_choice = input("Please select rock, paper, or scissors: ").lower().strip()

        if user_choice in options:
            return user_choice
        else:
            print("Invalid choice. Please choose rock, paper, or scissors.")
```

```python
def determine_winner(user_choice, computer_choice, winning_choices):
    """This function determines the winner of a round based on the user and computer choices.
    """
    if user_choice == computer_choice:
        return "tie"
    elif winning_choices[user_choice] == computer_choice:
        return "user"
    else:
        return "computer"
```

```python
def display_score(user, game_state):
    """This function determines the score after each round and displays it.
    """
    print(f"Score: {user}: {game_state['player_wins']} | Computer: {game_state['computer_wins']}")
```

```python
def ask_play_again():
    """
    Ask the player if they want to play another tournament.
    Return True if the player says yes and False if the player says no.
    """
    while True:
        play_again = input("Would you like to play again? (yes/no): ").lower().strip()

        if play_again == "yes":
            return True
        elif play_again == "no":
            return False
        else:
            print("Invalid input. Please enter yes or no.")
```

```python
while True:
    # Reset game state at the start of each new tournament
    game_state = {
        "current_round": 1,
        "player_wins": 0,
        "computer_wins": 0,
        "ties": 0,
        "round_history": []
    }

    # Continue playing until either the player or computer reaches wins_needed
    while game_state["player_wins"] < wins_needed and game_state["computer_wins"] < wins_needed:

        round_number = game_state["current_round"]

        print(f"\nRound {round_number}")

        user_choice = get_user_choice(options)
        computer_choice = get_computer_choice(options)

        print(f"{user} chose {user_choice}.")
        print(f"Computer chose {computer_choice}.")

        winner = determine_winner(user_choice, computer_choice, winning_choices)

        if winner == "tie":
            print("Tie match! No points awarded.")
            game_state["ties"] = game_state["ties"] + 1

        elif winner == "user":
            print(f"Congratulations {user}! You won this round!")
            game_state["player_wins"] = game_state["player_wins"] + 1
            game_state["current_round"] = game_state["current_round"] + 1

        else:
            print(f"Sorry {user}, the computer won this round.")
            game_state["computer_wins"] = game_state["computer_wins"] + 1
            game_state["current_round"] = game_state["current_round"] + 1

        # Store each round result in the game state dictionary
        game_state["round_history"].append({
            "round": round_number,
            "user_choice": user_choice,
            "computer_choice": computer_choice,
            "winner": winner
        })

        display_score(user, game_state)

    # Display the final tournament result
    if game_state["player_wins"] >= wins_needed:
        print(f"\nCongratulations {user}! You won the tournament!")
    elif game_state["computer_wins"] >= wins_needed:
        print(f"\nSorry {user}, you lost the tournament. Maybe try again?")

    # Ask if the player wants to play another tournament
    play_again = ask_play_again()

    if play_again == False:
        print("Thanks for playing!")
        break
```
